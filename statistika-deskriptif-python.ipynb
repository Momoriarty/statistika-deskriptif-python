```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. PENGORGANISASIAN DATA
path = "LatihanExcel_Arifin.xlsx"

dataraw = pd.read_excel(path, header=5, usecols="B:H")
dataraw = dataraw.dropna(how='all')

dataraw.columns = ["ID", "Name", "Quiz", "UTS", "UAS", "Final Score", "Grade"]

dataraw = dataraw.drop(0).reset_index(drop=True)

print("Data Awal (5 baris teratas):")
dataraw.head()
```

```python
# 2. TABEL FREKUENSI (Berdasarkan Grade)
datafrq = pd.crosstab(index=dataraw["Grade"], columns="Frekuensi")
print("Tabel Frekuensi Nilai:")
datafrq
```

```python
# 3A. Grafik Garis
plt.figure(figsize=(8, 5))
datafrq.plot(kind='line', color='dodgerblue', marker='o', linewidth=2)
plt.title('Grafik Garis - Distribusi Nilai')
plt.ylabel('Frekuensi')
plt.show()
```

```python
# 3B. Grafik Batang
plt.figure(figsize=(8, 5))
datafrq.plot(kind='bar', color='coral', edgecolor='black')
plt.title('Grafik Batang - Distribusi Nilai')
plt.ylabel('Frekuensi')
plt.xticks(rotation=0)
plt.show()
```

```python
# 3C. Pie Chart
plt.figure(figsize=(7, 7))
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0','#ffb3e6']
datafrq.plot(kind='pie', y='Frekuensi', autopct='%1.1f%%', legend=False, colors=colors, startangle=90)
plt.title('Pie Chart - Distribusi Nilai')
plt.ylabel('') # Menghilangkan label Y agar lebih rapi
plt.show()
```

```python
# 4. STATISTIKA DESKRIPTIF
# Note: Kalau nama kolom di excelmu 'Nilai', ganti kata 'Final Score' di bawah ini
dataraw["Final Score"] = pd.to_numeric(dataraw["Final Score"], errors='coerce')
dt = dataraw["Final Score"]

stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['variance'] = dt.var()
stats['range'] = dt.max() - dt.min()
stats['skewness'] = dt.skew()
stats['kurtosis'] = dt.kurtosis()

print("Ringkasan Statistika Deskriptif:")
stats
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Semua library berhasil diimpor!")
```

```python
data = {
    "Nama": ["Andi", "Budi", "Citra", "Dini", "Eko", "Fani", "Gilang", "Hana", "Indra", "Joko"],
    "Program_Studi": ["TI", "SI", "TI", "SI", "TI", "SI", "TI", "SI", "TI", "SI"],
    "Waktu_Belajar": [1, 2, 3, 4, 5, 2, 6, 3, 4, 5],
    "Kehadiran": [70, 75, 80, 85, 90, 75, 95, 80, 85, 90],
    "Nilai": [55, 60, 65, 78, 85, 62, 95, 70, 80, 90]
}

df = pd.DataFrame(data)
df
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    df["Nama"],
    df["Nilai"],
    color="skyblue"
)

plt.title("Nilai Ujian Mahasiswa")
plt.xlabel("Nama Mahasiswa")
plt.ylabel("Nilai Ujian")

plt.ylim(0, 100)
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.show()
```

```python
epoch = [1, 2, 3, 4, 5]
akurasi = [60, 68, 75, 82, 88]
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    epoch,
    akurasi,
    marker="o",
    color="green",
    linestyle="-"
)

plt.title("Perkembangan Akurasi Model")
plt.xlabel("Epoch")
plt.ylabel("Akurasi (%)")

plt.xticks(epoch)
plt.ylim(0, 100)
plt.grid(True, linestyle="--", alpha=0.4)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Nilai"],
    bins=5,
    color="orange",
    edgecolor="black"
)

plt.title("Distribusi Nilai Ujian Mahasiswa")
plt.xlabel("Nilai Ujian")
plt.ylabel("Frekuensi")

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Waktu_Belajar"],
    df["Nilai"],
    color="purple",
    s=80
)

plt.title("Hubungan Waktu Belajar dan Nilai Ujian")
plt.xlabel("Waktu Belajar (jam)")
plt.ylabel("Nilai Ujian")

plt.xlim(0, 7)
plt.ylim(0, 100)
plt.grid(True, linestyle="--", alpha=0.4)

plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="Waktu_Belajar",
    y="Nilai",
    hue="Program_Studi",
    s=100
)

plt.title("Waktu Belajar dan Nilai Berdasarkan Program Studi")
plt.xlabel("Waktu Belajar (jam)")
plt.ylabel("Nilai Ujian")

plt.show()
```

```python
data_numerik = df[
    ["Waktu_Belajar", "Kehadiran", "Nilai"]
]

korelasi = data_numerik.corr()

sns.heatmap(
    korelasi,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Heatmap Korelasi Variabel Mahasiswa")
plt.show()
```

```python
# Latihan

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('ecommerce_sales_analytics_5000.csv')

df.head()
```

```python
plt.figure(figsize=(10, 6))

data_bar = df.groupby('product_category')['revenue'].sum().sort_values(ascending=False)

plt.bar(data_bar.index, data_bar.values, color='skyblue')

plt.title('Total Pendapatan per Kategori Produk')
plt.xlabel('Kategori Produk (Product Category)')
plt.ylabel('Total Pendapatan (Revenue)')
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.show()
```

```python
plt.figure(figsize=(10, 5))

data_line = df.groupby('delivery_days')['customer_rating'].mean()

plt.plot(data_line.index, data_line.values, marker='o', color='orange', linewidth=2)

plt.title('Pengaruh Lama Pengiriman terhadap Rating Pelanggan')
plt.xlabel('Lama Pengiriman (Hari)')
plt.ylabel('Rata-rata Rating Pelanggan')
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

```python
plt.figure(figsize=(10, 5))

df['order_date'] = pd.to_datetime(df['order_date'])

data_line = df.groupby(df['order_date'].dt.year)['revenue'].sum()

plt.plot(data_line.index.astype(str), data_line.values, marker='o', color='green', linewidth=2)

plt.title('Tren Total Pendapatan per Tahun')
plt.xlabel('Tahun')
plt.ylabel('Total Pendapatan (Revenue)')
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='unit_price',
    y='revenue',
    hue='product_category',
    s=80,
    alpha=0.7
)

plt.title('Hubungan Harga Barang dengan Total Belanja')
plt.xlabel('Harga Barang Satuan')
plt.ylabel('Total Bayar (Revenue)')
plt.grid(True, linestyle='--', alpha=0.5)

plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title='Kategori Produk')
plt.tight_layout()

plt.show()
```
